```python
# 1. say_hi
def say_hi():
    print("Hello, world!")

# 2. goat_pad
def goat_pad(s):
    print("goat" + s + "goat")

# 3. print_n
def print_n(s, n):
    for _ in range(n):
        print(s)


# 测试
say_hi()
goat_pad("test")
print_n("hi", 3)
```

```python
#欧几里得算法
def gcd(m, n):
    while n != 0:
        m, n = n, m % n
    return m

# 测试三组数字
print(gcd(2018, 2019))   
print(gcd(1200, 300))    
print(gcd(5040, 60))     

# 测试负数情况
print(gcd(-12, -8))
print(gcd(-12, 8))
```

```python
#近似欧拉数e
import math

# 1. euler_limit
def euler_limit(n):
    return (1 + 1 / n) ** n

# 2. euler_infinite_sum
def euler_infinite_sum(n):
    if n == 0:
        return 0.0
    total = 0.0
    for k in range(n):  # k = 0,1,...,n-1，一共n项
        total += 1.0 / math.factorial(k)
    return total

# 3. euler_approx
def euler_approx(epsilon):
    true_e = math.exp(1)
    total = 0.0
    k = 0
    while abs(true_e - total) > epsilon:
        total += 1.0 / math.factorial(k)
        k += 1
    return total

# 4. print_euler_sum_table
def print_euler_sum_table(n):
    for k in range(1, n+1):
        val = euler_infinite_sum(k)
        print(val)


# 测试
print(euler_infinite_sum(1))  
print(euler_infinite_sum(4))  
print(euler_limit(10000))
print(euler_approx(0.001))
print_euler_sum_table(5)
```

```python
#测试整数属性
# 1. is_power_of_2
def is_power_of_2(n):
    while n > 1:
        if n % 2 != 0:  # 除以2有余数，不是2的幂
            return False
        n = n // 2
    return True

# 2. is_power(b, n)
def is_power(b, n):
    while n > 1:
        if n % b != 0:
            return False
        n = n // b
    return True


# 测试
print(is_power_of_2(8))   # True
print(is_power_of_2(10))  # False
print(is_power(3, 27))    # True
print(is_power(5, 30))    # False
```
