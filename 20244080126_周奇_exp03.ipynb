```python
#实验1 正则表达式基础匹配 任务1
# 原始文本：cat on mat; later: cats and dogs
# 规则说明：
# \d+   匹配1个或多个数字
# \s+   匹配1个或多个空白字符（空格、换行、tab）
# \bcat\b  \b是单词边界，只匹配独立单词cat，不会匹配cats里的cat
s1 = "cat on mat; later: cats and dogs"
print("原始文本：", s1)
print(r"\d+ 匹配：", re.findall(r"\d+", s1))
print(r"\s+ 匹配：", re.findall(r"\s+", s1))
print(r"\bcat\b 匹配：", re.findall(r"\bcat\b", s1))
```

```python
# 实验1 任务2
# 规则说明：
# cat|dog  | 代表或，匹配cat或者dog；flags=re.IGNORECASE 忽略大小写
# 正则分支特点：从左向右优先匹配左边分支
# r"a|aa|aaa"：优先匹配短的a
# r"aaa|aa|a"：优先匹配长的aaa
s2 = "cat dog bird CAT"
print("cat|dog（忽略大小写）匹配：", re.findall(r"cat|dog", s2, flags=re.IGNORECASE))
print('r"a|aa|aaa" 在 "aaaa"：', re.findall(r"a|aa|aaa", "aaaa"))
print('r"aaa|aa|a" 在 "aaaa"：', re.findall(r"aaa|aa|a", "aaaa"))
```

```python
#实验1任务3 量词对比 ==========
# 原始文本：ct cat caat caaat
# 量词规则：
# ca*t   * ：前面a出现0次或者多次
# ca+t   + ：前面a出现至少1次（1次及以上）
# ca?t   ? ：前面a出现0次或者1次
# ca{2}t {2}：前面a精确等于2次
# ca{1,2}t {1,2}：前面a出现1~2次（包含1和2）
s3 = "ct cat caat caaat"
print("ca*t  匹配：", re.findall(r"ca*t", s3))
print("ca+t  匹配：", re.findall(r"ca+t", s3))
print("ca?t  匹配：", re.findall(r"ca?t", s3))
print("ca{2}t 匹配：", re.findall(r"ca{2}t", s3))
print("ca{1,2}t 匹配：", re.findall(r"ca{1,2}t", s3))
```

```python
#实验2 分组、回溯引用与替换 任务1 提取邮箱用户名、域名
# 正则：(\w+)@([\w.]+)
# () 捕获分组：第1组(\w+) 匹配@前面用户名；第2组([\w.]+)匹配@后面域名
# \w 代表字母数字下划线
email_text = "联系：alice@example.com ; 备选：bob@test.org"
res_email = re.findall(r"(\w+)@([\w.]+)", email_text)
print("邮箱（用户名,域名）：", res_email)
```

```python
#实验2任务2 后向引用找连续重复单词 
# 正则：\b(\w+)\s+\1\b
# (\w+) 捕获一个单词存入分组1
# \1 是回溯引用，代表和分组1捕获内容一模一样，用来匹配重复单词
# \b 单词边界，防止部分单词误匹配
repeat_text = "cat cat, dog cat, bird bird"
res_repeat = re.findall(r"\b(\w+)\s+\1\b", repeat_text)
print("连续重复的单词：", res_repeat)
```

```python
#实验2任务3 日期转换：年-月-日 → 月-日-年
# 正则：(\d{4})[-/.](\d{2})[-/.](\d{2})
# 分组1：4位年份；分组2：2位月份；分组3：2位日期
# [-/.] 兼容横线、斜杠、点三种分隔符
# 替换表达式 \2-\3-\1：把 月、日、年 顺序重组
date_text = "2024-05-01\n2024/05/02\n2024.05.03"
print("=====转换前=====")
print(date_text)
date_pat = re.compile(r"(\d{4})[-/.](\d{2})[-/.](\d{2})")
date_new = date_pat.sub(r"\2-\3-\1", date_text)
print("=====转换后=====")
print(date_new)
```

```python
#实验2任务4 电话号码统一格式化
# 正则：(\d{3})[-.](\d{3})[-.](\d{4})
# 分组1：区号3位；分组2：中间3位；分组3：末尾4位
# [-.] 兼容横杠和点两种分隔符
# 替换式 r"(\1) \2-\3" 改成 (123) 456-7890 格式
phone_text = "联系我：123-456-7890 或 987.654.3210"
phone_pat = re.compile(r"(\d{3})[-.](\d{3})[-.](\d{4})")
phone_new = phone_pat.sub(r"(\1) \2-\3", phone_text)
print("格式化后电话：", phone_new)
```

```python
#实验3BeautifulSoup解析HTML 共用：定义HTML文本，构建解析树
html_doc = """
<html><head><title>实验课程</title></head><body>
<h1>数据分析</h1>
<p class="course">第六讲：正则表达式</p>
<p class="course">第七讲：结构化数据</p>
<a href="https://example.org/lecture06">讲义六</a>
<a href="https://example.org/lecture07">讲义七</a>
</body></html>
"""
# 使用python自带html.parser解析器，生成soup解析对象
soup = BeautifulSoup(html_doc, "html.parser")
```

```python
#实验3任务1：获取标题、h1标签
# find()：查找第一个匹配标签
# .name 读取标签名；.get_text() 提取标签内部纯文本，去掉标签符号
title_tag = soup.find("title")
h1_tag = soup.find("h1")
print("页面标题文本：", title_tag.get_text())
print("h1标签名：", h1_tag.name, "，h1文本：", h1_tag.get_text())
```

```python
#实验3任务2 find_all批量查找
# find_all()：查找所有满足条件的标签，返回列表
# class_="course" 筛选class属性等于course的p标签
# a["href"] 获取a标签里面href属性值
print("==== class='course' 的段落 ====")
p_list = soup.find_all("p", class_="course")
for p in p_list:
    print(p.get_text())

print("==== 所有链接 ====")
a_list = soup.find_all("a")
for a in a_list:
    print(f"链接文字：{a.get_text()}，href：{a['href']}")
```

```python
#实验3任务3 get_text()对比=
# soup.get_text()：提取整个文档所有可见文本
# 单个标签.get_text()：只提取这个标签内部文本
# repr() 用来显示换行、空白，方便对比差异
print("整个文档get_text：")
print(repr(soup.get_text()))
print("单个p标签get_text：", repr(p_list[0].get_text()))
print("单个a标签get_text：", repr(a_list[0].get_text()))
```
