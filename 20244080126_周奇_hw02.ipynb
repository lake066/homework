```python
def vec_scalar_mult(t, s):
    # 检查t是不是元组
    if not isinstance(t, tuple):
        raise TypeError("第一个参数必须是元组")
    # 检查标量s是不是int/float
    if not isinstance(s, (int, float)):
        raise TypeError("标量s必须是整数或者浮点数")
    # 检查元组里面每个元素都是数字
    for elem in t:
        if not isinstance(elem, (int, float)):
            raise TypeError("元组内元素只能是整数或者浮点数")
    # 逐元素相乘，返回新元组
    return tuple(x * s for x in t)

# 测试示例
print(vec_scalar_mult((1,2,3), 2))  # (2,4,6)
print(vec_scalar_mult((1.5, 2), 3)) # (4.5, 6)
```

```python
def vec_inner_product(v1, v2):
    if not isinstance(v1, tuple) or not isinstance(v2, tuple):
        raise TypeError("输入必须是元组")
    if len(v1) != len(v2):
        raise ValueError("两个向量长度不一致，无法计算内积")
    # 检查元素都是数字
    for elem in v1 + v2:
        if not isinstance(elem, (int, float)):
            raise TypeError("向量元素必须是整数或者浮点数")
    total = 0.0
    for a,b in zip(v1, v2):
        total += a * b
    return total

# 测试
print(vec_inner_product((1,2), (3,4))) # 1*3 +2*4 = 11.0
```

```python
def check_valid_mx(mx):
    # 外层必须是元组
    if not isinstance(mx, tuple):
        return False
    # 空矩阵也认为合法
    if len(mx) == 0:
        return True
    row_len = len(mx[0])
    for row in mx:
        # 每一行本身必须是元组
        if not isinstance(row, tuple):
            return False
        # 每行长度要一致
        if len(row) != row_len:
            return False
        # 每个元素是int/float
        for elem in row:
            if not isinstance(elem, (int, float)):
                return False
    return True

# 测试
print(check_valid_mx( ((1,2),(3,4)) )) # True
print(check_valid_mx( ((1,2),(3,4,5)) )) # False，行长度不一样
```

```python
def mx_vec_mult(mx, vec):
    # 先校验矩阵
    if not check_valid_mx(mx):
        raise TypeError("输入不是合法矩阵")
    if not isinstance(vec, tuple):
        raise TypeError("向量必须是元组")
    # 空矩阵特殊处理
    if len(mx) == 0:
        return ()
    n_cols = len(mx[0])
    if len(vec) != n_cols:
        raise ValueError("矩阵列数和向量维度不一致，无法相乘")
    # 检查向量元素是数字
    for e in vec:
        if not isinstance(e, (int, float)):
            raise TypeError("向量元素必须是数字")
    # 每行和vec求内积
    res = tuple( vec_inner_product(row, vec) for row in mx )
    return res

# 测试
M = ((1,2), (3,4))
v = (1,1)
print(mx_vec_mult(M, v)) # (3.0,7.0)
```

```python
#1 is_palindrome
def is_palindrome(s):
    clean = s.replace(" ", "").lower()
    return clean == clean[::-1]

print(is_palindrome("tacocat"))
print(is_palindrome("T A C O cat"))
print(is_palindrome("rats live on no evil star"))
```

```python
#is_abecedarian
def is_abecedarian(s):
    clean = s.replace(" ", "").lower()
    for i in range(len(clean)-1):
        if clean[i] > clean[i+1]:
            return False
    return True

print(is_abecedarian("adder"))
print(is_abecedarian("abcd efgh xyz"))
print(is_abecedarian("dog"))
```

```python
#remove_vowels
def remove_vowels(s):
    vowels = {"a", "e", "i", "o", "u"}
    out = []
    for ch in s:
        if ch.lower() not in vowels:
            out.append(ch)
    return "".join(out)

print(remove_vowels("cat"))
print(remove_vowels("audio"))
```

```python
#list_reverse
def list_reverse(lst):
    if not isinstance(lst, list):
        raise TypeError("输入必须是列表")
    return lst[::-1]

print(list_reverse([1, 2, 3]))
```

```python
#is_sorted
def is_sorted(seq):
    for i in range(len(seq)-1):
        if seq[i] > seq[i+1]:
            return False
    return True

print(is_sorted([1,2,3,4]))
print(is_sorted([1,3,2]))
```

```python
#vec_scalar_mult
def vec_scalar_mult(t, s):
    if not isinstance(t, tuple):
        raise TypeError("第一个参数必须是元组")
    if not isinstance(s, (int, float)):
        raise TypeError("标量s必须是整数或者浮点数")
    for elem in t:
        if not isinstance(elem, (int, float)):
            raise TypeError("元组内元素只能是整数或者浮点数")
    return tuple(x * s for x in t)

print(vec_scalar_mult((1,2,3), 2))
```

```python
#vec_inner_product
def vec_inner_product(v1, v2):
    if not isinstance(v1, tuple) or not isinstance(v2, tuple):
        raise TypeError("输入必须是元组")
    if len(v1) != len(v2):
        raise ValueError("两个向量长度不一致，无法计算内积")
    for elem in v1 + v2:
        if not isinstance(elem, (int, float)):
            raise TypeError("向量元素必须是整数或者浮点数")
    total = 0.0
    for a,b in zip(v1, v2):
        total += a * b
    return total

print(vec_inner_product((1,2), (3,4)))
```

```python
#check_valid_mx
def check_valid_mx(mx):
    if not isinstance(mx, tuple):
        return False
    if len(mx) == 0:
        return True
    row_len = len(mx[0])
    for row in mx:
        if not isinstance(row, tuple):
            return False
        if len(row) != row_len:
            return False
        for elem in row:
            if not isinstance(elem, (int, float)):
                return False
    return True

print(check_valid_mx( ((1,2),(3,4)) ))
```

```python
#mx_vec_mult
def mx_vec_mult(mx, vec):
    if not check_valid_mx(mx):
        raise TypeError("输入不是合法矩阵")
    if not isinstance(vec, tuple):
        raise TypeError("向量必须是元组")
    if len(mx) == 0:
        return ()
    n_cols = len(mx[0])
    if len(vec) != n_cols:
        raise ValueError("矩阵列数和向量维度不一致，无法相乘")
    for e in vec:
        if not isinstance(e, (int, float)):
            raise TypeError("向量元素必须是数字")
    res = tuple( vec_inner_product(row, vec) for row in mx )
    return res

M = ((1,2), (3,4))
v = (1,1)
print(mx_vec_mult(M, v))
```

```python
#vector类init
class Vector:
    def __init__(self, dim, elements=None):
        if not isinstance(dim, int):
            raise TypeError("维度必须是整数")
        if dim < 0:
            raise ValueError("维度不能为负数")
        if elements is None:
            elements = [0.0] * dim
        elements = tuple(elements)
        if len(elements) != dim:
            raise ValueError("元素数量和维度dim不一致")
        for e in elements:
            if not isinstance(e, (int, float)):
                raise TypeError("向量元素只能是整数/浮点数")
        self.dim = dim
        self.elements = elements
```

```python
#比较运算符
# 给Vector增加比较方法
def __eq__(self, other):
    if not isinstance(other, Vector):
        return False
    if self.dim != other.dim:
        raise ValueError("不能比较维度不同的Vector")
    return self.elements == other.elements

def __lt__(self, other):
    if not isinstance(other, Vector):
        raise TypeError("只能和Vector对象比较")
    if self.dim != other.dim:
        raise ValueError("不能比较维度不同的Vector")
    return self.elements < other.elements

def __le__(self, other):
    return self < other or self == other

def __gt__(self, other):
    return not (self <= other)

def __ge__(self, other):
    return not (self < other)

Vector.__eq__ = __eq__
Vector.__lt__ = __lt__
Vector.__le__ = __le__
Vector.__gt__ = __gt__
Vector.__ge__ = __ge__

# 测试
v1 = Vector(2, (2,4))
v2 = Vector(2, (2,5))
print(v1 < v2)
print(v1 <= v2)
print(v1 > v2)
print(v1 >= v2)
print(v1 == v2)
```

```python
#dot内积方法
def dot(self, other):
    if not isinstance(other, Vector):
        raise TypeError("dot的参数必须是Vector实例")
    if self.dim != other.dim:
        raise ValueError("两个向量维度不一致，不能计算内积")
    s = 0.0
    for a,b in zip(self.elements, other.elements):
        s += a*b
    return s

Vector.dot = dot

# 测试
v1 = Vector(2, (2,4))
v2 = Vector(2, (2,5))
print(v1.dot(v2))
```

```python
#乘法运算符
def __mul__(self, other):
    if isinstance(other, (int, float)):
        new_elems = tuple(x * other for x in self.elements)
        return Vector(self.dim, new_elems)
    elif isinstance(other, Vector):
        if self.dim != other.dim:
            raise ValueError("相乘的向量维度不一致")
        new_elems = tuple(a*b for a,b in zip(self.elements, other.elements))
        return Vector(self.dim, new_elems)
    else:
        raise TypeError("不支持该类型相乘")

def __rmul__(self, other):
    return self * other

Vector.__mul__ = __mul__
Vector.__rmul__ = __rmul__

# 测试
v1 = Vector(2, (2,4))
print(2 * v1)
print(v1 * v1)
```
