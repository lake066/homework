```python
#实验二 分数评分函数
def score_level(score):
    if not(0 <= score <=100):
        raise ValueError("分数必须在0~100之间")
    if 90 <= score <=100:
        return "优秀"
    elif 80 <= score <=90:
        return "良好"
    elif 60 <= score <80:
        return "及格"
    else:
        return "不及格"
#测试
print(score_level(95))
print(score_level(72))
```

```python
#递归版
def fact_rec(n):
    if n==0:
        return 1
    return n*fact_rec(n-1)
#迭代版
def fact_iter(n):
    res = 1
    for i in range(1,n+1):
        res *=i
    return res
#测试
for n in [0,1,5,10]:
    print(f"n={n},递归：{fact_rec(n)},迭代: {fact_iter(n)}")
```

```python
#返回最大，最小，平均值
def summary(values):
    minv = min(values)
    maxv = max(values)
    avg = sum(values)/len(values)
    return minv ,maxv, avg

res = summary([85,58,92,76,60,89])
min_s, max_s, avg_s = res
print(min_s, max_s, avg_s)
```

```python
import os
import pickle

#查看工作目录、构造路径、创建exp01_data文件夹
print("当前工作目录：", os.getcwd())
data_dir = os.path.join(os.getcwd(), "exp01_data")
os.makedirs(data_dir, exist_ok=True)

students_txt_path = os.path.join(data_dir, "students.txt")
report_path = os.path.join(data_dir, "report.txt")
pkl_path = os.path.join(data_dir, "students.pkl")
```

```python
#写入txt原始学生数据
raw_text = """name,score
Alice,85
Bob,58
Carol,92
David,76
Eva,60
Frank,89
Grace,abc
Henry,105
Incomplete
"""
with open(students_txt_path, "w", encoding="utf-8") as f:
    f.write(raw_text)
print("students.txt 写入完成")
```

```python
#读取文件，跳过表头，循环解析+捕获异常
from student_utils import parse_student

valid_records = []
error_log = []

with open(students_txt_path, "r", encoding="utf-8") as f:
    lines = f.readlines()


for idx, line in enumerate(lines[1:], start=2):
    try:
        rec = parse_student(line)
        valid_records.append(rec)
    except ValueError as e:
        error_log.append( (idx, str(e)) )

# 统计指标
valid_cnt = len(valid_records)
invalid_cnt = len(error_log)
if valid_cnt > 0:
    score_list = [d["score"] for d in valid_records]
    avg_score = sum(score_list)/len(score_list)
    pass_cnt = sum(1 for s in score_list if s >=60)
    pass_rate = pass_cnt / valid_cnt
else:
    avg_score = 0
    pass_rate = 0

# 打印结果
print("统计结果")
print(f"有效记录：{valid_cnt} 条")
print(f"无效记录：{invalid_cnt} 条")
print(f"平均分：{avg_score:.2f}")
print(f"及格率：{pass_rate:.2%}")
print("无效记录详情：")
for row, msg in error_log:
    print(f"第{row}行：{msg}")
```

```python
# W模式，a模式
content_w = f"""【w模式写入】
有效记录数：{valid_cnt}
无效记录数：{invalid_cnt}
平均分：{avg_score:.2f}
及格率：{pass_rate:.2%}
------------------------
"""
# w：覆盖原有文件
with open(report_path, "w", encoding="utf-8") as f:
    f.write(content_w)

# a：追加，不覆盖
content_a = f"""【a模式追加的内容】
追加时间演示，a模式不会清空前面内容
========================
"""
with open(report_path, "a", encoding="utf-8") as f:
    f.write(content_a)

print("report.txt 已生成，w覆盖、a追加演示完成")
```

```python
#pickle序列化、反序列化
with open(pkl_path, "wb") as f:
    pickle.dump(valid_records, f)

# 读取回来校验
with open(pkl_path, "rb") as f:
    loaded_data = pickle.load(f)

print("pickle读取校验，和原数据相等？", loaded_data == valid_records)
```

```python
#实验五
import copy

# 任务1：Student类
class Student:
    def __init__(self, name, score):
        self.name = name
        self.score = score
    def is_passed(self):
        return self.score >= 60
    def __str__(self):
        status = "及格" if self.is_passed() else "不及格"
        return f"Student(name={self.name}, score={self.score}, {status})"
```

```python
#批量生成Student对象，打印
print("Student对象列表")
stu_objs = []
for d in valid_records:
    s = Student(d["name"], d["score"])
    stu_objs.append(s)
    print(s)
```

```python
# 任务3：继承 Student 的研究生类
class GraduateStudent(Student):
    def __init__(self, name, score, research_topic):
        super().__init__(name, score)
        self.research_topic = research_topic
    def __str__(self):
        base_msg = super().__str__()
        return f"{base_msg}, 研究方向：{self.research_topic}"

grad1 = GraduateStudent("LiMing", 88, "大数据挖掘")
print("\n==== 研究生子类示例 ====")
print(grad1)
```

```python
# 任务4 拷贝对比实验
print("\n==== 拷贝对比实验 ====")

# ①直接赋值
orig = {"name": "Alice", "scores": [85, 90]}
obj_assign = orig
obj_assign["scores"].append(99)
print("直接赋值后 orig：", orig)

# 重置原始对象
orig = {"name": "Alice", "scores": [85, 90]}
# ②浅拷贝 copy.copy
obj_shallow = copy.copy(orig)
obj_shallow["scores"].append(99)
print("浅拷贝后 orig：", orig)

# 重置原始对象
orig = {"name": "Alice", "scores": [85, 90]}
# ③深拷贝 deepcopy
obj_deep = copy.deepcopy(orig)
obj_deep["scores"].append(99)
print("深拷贝后 orig：", orig)
```
